```python
from IPython.display import Image, Markdown, display

from umkm import config, eda
from umkm.eda_teks import interpretasi

rep = eda.run()
reports = config.path("reports")
print(f"{len(rep.tabel)} tabel, {len(rep.grafik)} grafik")
```

## Ringkasan temuan

```python
display(Markdown((reports / "eda_ringkasan.md").read_text(encoding="utf-8")))
```

## Grafik dan interpretasi

```python
import pandas as pd

df = pd.read_csv(config.path("processed") / "data_features.csv")
interp = interpretasi(rep.tabel, df)
for nama, path in rep.grafik.items():
    display(Markdown(f"### {nama}"))
    display(Image(filename=str(path)))
    display(Markdown(f"**Interpretasi:** {interp[nama]}"))
```
