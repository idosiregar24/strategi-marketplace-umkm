Notebook ini hanya memanggil `umkm.pengaruh.run()` lalu menampilkan hasilnya. Logika ada di
`src/umkm/pengaruh.py`, `models.py`, `stats.py`, dan `pengaruh_teks.py`. Jalankan ulang lewat CLI:
`python -m umkm.pipeline --stage 4` (bootstrap 1000x, sekitar 5-6 menit).

Keluaran: `reports/tables/pengaruh_*.csv`, `reports/figures/pengaruh_*.png` dan `shap_*.png`,
`reports/pengaruh_log.md`, `reports/pengaruh_ringkasan.md`. Semua hasil bersifat asosiasi,
bukan sebab-akibat.

```python
import warnings

from IPython.display import Image, Markdown, display

from umkm import config, pengaruh
from umkm.pengaruh_teks import interpretasi

warnings.filterwarnings("ignore")
rep = pengaruh.run()
reports = config.path("reports")
print(f"{len(rep.tabel)} tabel, {len(rep.grafik)} grafik, bootstrap {rep.info['n_boot']}x")
```

## Ringkasan temuan

```python
display(Markdown((reports / "pengaruh_ringkasan.md").read_text(encoding="utf-8")))
```

## Grafik dan interpretasi

```python
interp = interpretasi(rep)
for nama, path in rep.grafik.items():
    display(Markdown(f"### {nama}"))
    display(Image(filename=str(path)))
    display(Markdown(f"**Interpretasi:** {interp[nama]}"))
```
