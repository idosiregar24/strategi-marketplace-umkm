# Tahap 5: Segmentasi kondisi UMKM

Notebook ini hanya memanggil `umkm.segmentation.run()` lalu menampilkan hasilnya. Logika ada di
`src/umkm/segmentation.py` dan `segmentasi_teks.py`. Jalankan ulang lewat CLI:
`python -m umkm.pipeline --stage 5` (sekitar 40 detik).

Segmen dibentuk hanya dari profil UMKM (skala omzet, lama usaha sebelum MP, lama di MP, jumlah
platform; D27). Keluaran: `data/processed/data_segmen.csv`, `reports/tables/profil_segmen.csv` dan
`segmen_*.csv`, `reports/figures/segmen_*.png`, `reports/segmentasi_log.md`,
`reports/segmentasi_ringkasan.md`.

```python
import warnings

from IPython.display import Image, Markdown, display

from umkm import config, segmentation
from umkm.segmentasi_teks import interpretasi

warnings.filterwarnings("ignore")
rep = segmentation.run()
reports = config.path("reports")
print(f"k = {rep.info['k']}, silhouette {rep.info['silhouette']:.3f}, {len(rep.grafik)} grafik")
```

## Ringkasan

```python
display(Markdown((reports / "segmentasi_ringkasan.md").read_text(encoding="utf-8")))
```

## Grafik dan interpretasi

```python
interp = interpretasi(rep)
for nama, path in rep.grafik.items():
    display(Markdown(f"### {nama}"))
    display(Image(filename=str(path)))
    display(Markdown(f"**Interpretasi:** {interp[nama]}"))
```
